```python
areas_df = df[['city', 'locality']].dropna().drop_duplicates()

print(areas_df.shape)
areas_df.head()
```

```python
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut
import pandas as pd
import time


geolocator = Nominatim(user_agent="houseiq")


coordinates = []


for _, row in areas_df.iterrows():

    city = row['city']
    locality = row['locality']

    query = f"{locality}, {city}, India"

    try:

        location = geolocator.geocode(
            query,
            timeout=10
        )


        if location:

            coordinates.append({

                'city': city,

                'locality': locality,

                'latitude': location.latitude,

                'longitude': location.longitude

            })

            print(f"✓ {query}")


        else:

            coordinates.append({

                'city': city,

                'locality': locality,

                'latitude': None,

                'longitude': None

            })

            print(f"✗ {query}")


        time.sleep(1)


    except GeocoderTimedOut:


        coordinates.append({

            'city': city,

            'locality': locality,

            'latitude': None,

            'longitude': None

        })

        print(f"Timeout : {query}")



coords_df = pd.DataFrame(coordinates)

coords_df.to_csv(

    "data/coordinates.csv",

    index=False

)

print("Saved Successfully")
```
